```python
payment_methods = [
    "UPI",
    "Debit Card",
    "Credit Card",
    "Net Banking"
]
```

```python
amount_range = {

    "Food": (150,1500),

    "Travel": (800,12000),

    "Transport": (80,1200),

    "Online Shopping": (300,10000),

    "Electronics": (3000,120000),

    "Healthcare": (150,6000),

    "Bills": (300,8000),

    "Entertainment": (200,3000),

    "Grocery": (300,5000),

    "Clothing": (500,8000)

}
```

```python
def random_date():
    
    start = datetime(2025,1,1)
    
    end = datetime(2025,12,31)
    
    delta = end - start
    
    random_days = random.randint(0, delta.days)
    
    return start + timedelta(days=random_days)
```

```python
new_rows = []

for i in range(500):

    category = random.choice(list(merchant_data.keys()))

    description = random.choice(merchant_data[category])

    amount = round(
        random.uniform(
            amount_range[category][0],
            amount_range[category][1]
        ),
        2
    )

    payment = random.choice(payment_methods)

    date = random_date()

    new_rows.append({

        "Date": date.strftime("%Y-%m-%d"),

        "Description": description.lower(),

        "Amount": amount,

        "Category": category,

        "PaymentMethod": payment,

        "Month": date.strftime("%B"),

        "Year": date.year,

        "Month_Number": date.month,

        "Day": date.day,

        "Weekday": date.strftime("%A")

    })
```

```python
new_df = pd.DataFrame(new_rows)

new_df.head()
```

```python
augmented_df = pd.concat(
    [df, new_df],
    ignore_index=True
)
```
